## 2. Discover a real truth yourself

The safest, most satisfying early wins are **things you find in the data** — un-leakable, and they *are* the core lesson. Run the three cells below. Each is ~10 lines of pandas and each overturns a common SEO belief.

Every number is **computed live from the shipped CSV** — nothing is hardcoded.

```python
import pandas as pd, numpy as np
df = pd.read_csv("data/raw/content_refresh_anonymized.csv")
print(df.shape[0], "rows,", df.shape[1], "columns")
df.head(3)
```

### Discovery A — "High search volume means more traffic." Does it?

```python
corr = df["search_volume"].corr(df["impressions_90d"])
print(f"Correlation between search_volume and impressions_90d: {corr:.3f}")
print("Near zero -> keyword search volume barely predicts the traffic a page actually gets.")
```

### Discovery B — the CTR cliff by position
Click-through rate is not flat: it collapses as you move down the results.

```python
visible = df[df["impressions_90d"] >= 100]
ctr_by_pos = visible.groupby("position_tier")["ctr"].mean().sort_values(ascending=False)
print(ctr_by_pos.round(4).to_string())
ctr_by_pos.plot(kind="bar", title="Mean CTR by position tier (impressions >= 100)", ylabel="CTR");
```

### Discovery C — is longer content the lever?
Compare word count for **declining** vs **growing** pages.

```python
wc = df.groupby("trend_direction")["word_count"].median()
print(wc.round(0).to_string())
print("\n'down' vs 'up' pages have almost the same median word count -> length is not the lever.")
```

```python
# Your discovery here
df_filtered = df[df["impressions_90d"] > 0]

corr = df_filtered["search_volume"].corr(df_filtered["impressions_90d"])
print(f"Correlation between search_volume and impressions_90d (only impressions_90d > 0): {corr:.3f}")

if pd.isna(corr):
    print("Correlation is undefined (likely because one of the columns has no variation after filtering).")
elif abs(corr) < 0.1:
    print("Near zero -> keyword search volume barely predicts the traffic a page actually gets.")
elif corr > 0:
    print("Positive correlation -> higher search volume tends to be associated with more impressions.")
else:
    print("Negative correlation -> higher search volume tends to be associated with fewer impressions.")
```

### **Does the correlation change?**
No , the correlation is still 0.001 .Restricting the analysis to pages with impressions_90d > 0 produced a correlation of 0.001 between search_volume and impressions_90d. The correlation remains near zero, so keyword search volume is still not a meaningful predictor of the impressions a page receives.

```python
visible = df[df["impressions_90d"] >= 100]

ctr_by_content = visible.groupby("content_type")["ctr"].mean().sort_values()

print(ctr_by_content.round(4).to_string())

worst_content = ctr_by_content.index[0]
worst_ctr = ctr_by_content.iloc[0]

print(f"\nWorst content type: {worst_content} (CTR = {worst_ctr:.4f})")
ctr_by_content.plot(kind="bar", title="Mean CTR by content tier (impressions >= 100)", ylabel="CTR");
```

### **In Discovery B, which content_type has the worst CTR at the same position tier?**
Comparison Article has the lowest mean CTR at 0.1340, making it the worst-performing content type in this comparison. Keyword Article performs better at around 0.255, while Feedly Article has by far the highest CTR at around 0.709.

```python
age_by_trend = df.groupby("trend_direction")["content_age_days"].median()

print(age_by_trend.round(1).to_string())

age_by_trend.plot(
    kind="bar",
    title="Average Content Age by Trend Direction",
    ylabel="Content Age (Days)"
);
corr = df["avg_position"].corr(df["ctr"])
print(f"Correlation between avg_position and ctr: {corr:.3f}")
```

###**Another belief to test: does `content_age_days` relate to `trend_direction`? Does `avg_position` relate to `ctr`**

Older content appears more likely to have **stable or upward trends**. Pages with **stable** and **up** trend directions have the highest **median** content age (300.0 and 291.5 days), while **down** and **flat** pages have lower median ages (216.0 and 231.0 days). Additionally, the correlation between `avg_position` and `ctr` is **-0.073**, indicating only a very weak negative relationship. This suggests that content age may be related to trend direction, while average position shows little association with CTR in this dataset.
